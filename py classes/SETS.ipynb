# Sets in Python

A **set** is a collection where every item is **unique**. We create it with curly brackets `{ }` and separate the items with commas `,`

---

| Property | Set | What it means |
|---|---|---|
| Ordered | No | Python decides the order, we cannot rely on it |
| Indexed | No | No `marks[0]`, there is no position number |
| Changeable | Yes, through methods | We can add or remove items, but cannot replace one by index |
| Duplicates | Not allowed | Same value given twice is stored only once |

**Real life:** think of the attendance register. Even if a student's name is written twice by mistake, the student is counted only once. A set works the same way.

## 1. Creating a set

```python
# Creating a set of marks using curly brackets { }
# Note that 12 is written four times on purpose
marks = {12, 43, 5456, 23, 2, 12, 12, 12}
```

```python
# type() tells us the data type, here it will say 'set'
# len() counts the items; duplicates of 12 are removed, so we get 5 and not 8
print(type(marks), len(marks))
```

```python
# Printing the set: order may look different from what we typed (unordered)
# and 12 comes only once (no duplicates allowed)
print(marks)
```

### Empty set: a common mistake
`{}` alone creates an empty **dictionary**, not a set. For an empty set always write `set()`.

```python
empty_one = {}          # this is a dictionary, not a set
empty_two = set()       # this is the correct way to make an empty set

print(type(empty_one))  # <class 'dict'>
print(type(empty_two))  # <class 'set'>
```

### Removing duplicates from a list
Passing a list to `set()` keeps only the unique values. This is the most common real-life use of sets.

```python
# Cities from which orders came today, some cities are repeated
order_cities = ["Delhi", "Mumbai", "Delhi", "Pune", "Mumbai", "Jaipur"]

unique_cities = set(order_cities)   # set() removes the repeated city names
print(unique_cities)
print(len(unique_cities), "unique cities")   # count of different cities
```

## 2. Sets are unindexed and cannot be changed by position

```python
# Trying to read the first item by index
# Sets have no index, so Python raises TypeError: 'set' object is not subscriptable
marks[0]
```

```python
# Trying to replace an item by index
# This also fails, because a set has no positions to assign to
# (We change a set only through methods like add, remove, update)
marks[0] = 10000
```

### Checking membership with `in`
Sets cannot be indexed, but checking whether a value exists is very fast.

```python
print(43 in marks)        # True, 43 is present in the set
print(100 in marks)       # False, 100 is not present
print(100 not in marks)   # True, 'not in' is the opposite check
```

```python
# We can still loop over a set; each item comes once, order is not fixed
for m in marks:
    print(m)
```

## 3. Adding items

```python
# add() puts ONE new item into the set
marks.add(999)
print(marks)

# Adding an item that already exists does nothing, no error either
marks.add(999)
print(marks)
```

```python
# update() adds MANY items at once
# We can pass a list, a set or a tuple; each item inside gets added
marks.update({899, 10000, 567})
print(marks)
```

## 4. Removing items

```python
# pop() removes and returns ONE item
# Since a set is unordered, we cannot know in advance which item will go
removed = marks.pop()
print("Removed:", removed)
print(marks)
```

```python
# remove() deletes the item we mention
marks.remove(5456)
print(marks)
```

```python
# remove() raises KeyError if the item is not in the set
marks.remove(987654)
```

```python
# discard() also deletes the item we mention,
# but if the item is not found it quietly does nothing (no error)
marks.discard(1999999)
print(marks)
```

> **Tip:** use `remove()` when the item *must* be there (an error helps you catch a bug).
> Use `discard()` when you are not sure and just want it gone.

## 5. Set operations (maths of sets)

We will use these two sets for every example below:

- `a = {1, 2, 3, 4, 5}`
- `b = {3, 4, 5, 6, 7}`

Each method has a **normal** version (returns a new set, `a` stays the same) and an **`_update`** version (changes `a` itself and returns `None`).

```python
a = {1, 2, 3, 4, 5}
b = {3, 4, 5, 6, 7}
```

### Intersection: items common to both

```python
# intersection() returns a NEW set with the items present in both a and b
# Short form: a & b
print(a.intersection(b))
print(a & b)
print(a)   # a is unchanged
```

```python
# intersection_update() changes 'a' itself so that it keeps only the common items
a = {1, 2, 3, 4, 5}          # fresh copy of a for this example
a.intersection_update(b)     # now a = {3, 4, 5}
print(a)
```

### Difference: items in `a` but not in `b`

```python
a = {1, 2, 3, 4, 5}
# difference() returns a NEW set with items that are in a but NOT in b
# Short form: a - b
print(a.difference(b))
print(a - b)
print(b - a)   # order matters: items in b but not in a
```

```python
a = {1, 2, 3, 4, 5}
# difference_update() changes 'a' itself; it returns None, so we print a separately
print(a.difference_update(b))   # prints None
print(a)                        # now a = {1, 2}
```

### Symmetric difference: items that are NOT common

```python
a = {1, 2, 3, 4, 5}
# symmetric_difference() returns items that are in a OR b, but not in both
# Short form: a ^ b
print(a.symmetric_difference(b))
print(a ^ b)
```

```python
a = {1, 2, 3, 4, 5}
# symmetric_difference_update() changes 'a' itself; returns None
print(a.symmetric_difference_update(b))   # prints None
print(a)                                  # now a = {1, 2, 6, 7}
```

### Union: all items from both, without repeats

```python
a = {1, 2, 3, 4, 5}
# union() returns a NEW set with every item from a and b; common items come once
# Short form: a | b
x = a.union(b)
print(x)
print(a | b)
```

## 6. Comparing sets (answers are True / False)

```python
# isdisjoint() returns True if the two sets have NO item in common
print(x.isdisjoint(marks))       # True, x and marks share nothing
print(x.isdisjoint({7, 8, 9}))   # False, 7 is common
```

```python
y = {1, 2, 3, 4}
print(x)
print(y)
```

```python
# issuperset() returns True if x contains ALL the items of y
# Short form: x >= y
print(x.issuperset(y))
print(x >= y)
```

```python
# issubset() returns True if ALL items of y are present in x
# Short form: y <= x
print(y.issubset(x))
print(y <= x)
print(x.issubset(y))   # False, x has items (5, 6, 7) that y does not have
```

## 7. Copy and clear

```python
# copy() makes a separate new set with the same items
z = y.copy()
z.add(100)    # changing z does not affect y
print("y:", y)
print("z:", z)
```

```python
# clear() removes every item, the set becomes empty: set()
z.clear()
print(z)
print(len(z))
```

## Quick summary

| Method | Short form | Changes original? | Result |
|---|---|---|---|
| `add(x)` | | Yes | adds one item |
| `update(iterable)` | | Yes | adds many items |
| `pop()` | | Yes | removes any one item |
| `remove(x)` | | Yes | removes x, KeyError if missing |
| `discard(x)` | | Yes | removes x, no error if missing |
| `union(b)` | `a \| b` | No | all items |
| `intersection(b)` | `a & b` | No | common items |
| `difference(b)` | `a - b` | No | in a, not in b |
| `symmetric_difference(b)` | `a ^ b` | No | not common |
| `..._update(b)` | `&=`, `-=`, `^=` | Yes | same as above, saved into a |
| `isdisjoint(b)` | | No | True if nothing common |
| `issubset(b)` | `a <= b` | No | True if all of a is in b |
| `issuperset(b)` | `a >= b` | No | True if a has all of b |
| `copy()` | | No | new separate set |
| `clear()` | | Yes | empties the set |
