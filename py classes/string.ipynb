# Data Types

Every value we store in Python has a "type". The type tells Python what kind of value it is, like a number, a word or a yes/no answer. Python decides how to treat the value based on this type.

Think of it like your kitchen containers. Sugar goes in one box, oil goes in a bottle. Same way, each kind of data has its own type.

- `int` (integer) : a number without decimal point. Example: `10`, `-5`, `2024`
- `float` : a number with decimal point. Example: `3.14`, `99.5`
- `bool` / boolean : only two values are possible, `True` or `False`. Like a yes/no answer. (Note the capital T and F, Python is strict about it)
- `str` (string) : text. It can have alphabets, numbers, special symbols or even spaces, but it must be written inside single quotes `' '` or double quotes `" "`. Example: `"hello"`, `'Rohan123'`
- `None` / `NoneType` : means "nothing" or "no value". It is not zero and not an empty string, it simply means there is no value at all.
- `collections` (we will study these in later classes)
  - `list`
  - `set`
  - `tuple`
  - `dictionary`

##### Note: `type()` function is used to find the data type of any value or variable.
Just put the value inside the brackets and Python will tell you its type.

## String
A string is a group of characters (`alphabets, numbers, special symbols or whitespace`) written inside single `' '` or double `" "` quotes.

Important point: if something is inside quotes, Python treats it as text, even if it looks like a number. So `"1.23"` is a string, but `1.23` is a float.

````python
# print() shows the result on the screen, type() tells the data type of the value inside it
print(type(123))     # whole number, so the answer is int
print(type(1.23))    # number with a decimal point, so the answer is float
print(type("1.23"))  # looks like a number, but it is inside quotes, so Python says str (string)
print(type(True))    # True is a boolean value, so bool
print(type(False))   # False is also a boolean value, so bool again
print(type(None))    # None means no value, its type is NoneType
````

#### Note: Whatever the user types using the `input()` function always comes as a `str` (string). 
Even if the user types a number like `45`, Python gets it as text `"45"`. This is a very common confusion for beginners, so remember it well.

````python
# input() shows the message to the user, waits for them to type, and gives back what they typed
# the typed answer is saved in the variable on the left side of the = sign
name = input("Enter your name : ")
age = input("Enter your age : ")
isMale = input("Are you male ? : ")

# print() can show more than one thing at once if we separate them with a comma
# here we print the value first and then its type
print(name, type(name))      # john is text, so str
print(age, type(age))        # 45 looks like a number, but input() gave it as str
print(isMale, type(isMale))  # True also came as plain text, so str, not bool
````

#### Note : `len()` function tells us how many characters are there in a string.
`len` is short for length. Spaces and symbols are also counted as characters.

````python
# these variables came from the input cell above (john, 45, True)
print(len(name))    # "john" has 4 characters
print(len(age))     # "45" has 2 characters
print(len(isMale))  # "True" has 4 characters
````

````python
# creating a variable named txt and storing a string in it
txt = "hello world"
print(txt)        # shows the text
print(len(txt))   # 11, because h-e-l-l-o is 5, the space is 1 and w-o-r-l-d is 5
print(type(txt))  # str
````

````python
# a string can have letters, numbers and symbols like @ and . all together
username = "rubi123@email.com"
````

````python
print(username)        # prints the email text
print(len(username))   # counts every character, including @ and the dot, total 17
print(type(username))  # still a str, even though it has numbers and symbols inside
````

### Concatenation
Concatenation means joining two or more strings together to make one bigger string. We use the `+` sign for this.

Example: `"good" + "morning"` gives `"goodmorning"`. Notice that no space is added by itself, we have to add the space ourselves, like `"good " + "morning"`.

````python
# taking three inputs from the user, all three will be strings
name = input("Enter name : ")
city = input("Enter city : ")
age = input("Enter age : ")
````

````python
# joining small pieces of text and our variables using +
# see the spaces inside the quotes, like "my name is ", they are needed or words will stick together
# this works because name, age and city are all strings
print("my name is "+ name + " and i am " + age + " years old living in " + city)
````

````python
name = input("Enter name : ")
city = input("Enter city : ")
# int() changes the typed text into a real whole number
# the input() part runs first, and its answer goes inside int()
age = int(input("Enter age : "))  # now age is an integer, not a string
````

````python
# checking the type, it is int now because we used int() above
print(type(age))
````

````python
# This cell gives an ERROR, and it is on purpose, to show you what goes wrong.
# The + sign can join text with text only. Here age is a number (int), so Python gets confused.
# It is like trying to stick a number onto a sentence, Python does not know how to do it.
print("my name is "+ name + " and i am " + age + " years old living in " + city)
````

# Formatted string (f-string)
An f-string is written as `f""` or `f''`. We just put the letter `f` before the opening quote.

Inside it, whatever we write in curly brackets `{ }` is replaced by the actual value of that variable or expression.

Why it is better than `+` :
- no need to keep opening and closing quotes again and again
- no need to convert numbers into strings, so the TypeError we saw above will not come
- the code is easy to read

````python
# same sentence as before, but now with an f-string
# {name}, {age} and {city} get replaced by their values
# age is an int here, and still there is no error, that is the benefit of f-string
print(f"my name is {name} and i am {age} years old living in {city}")
````

### String methods
A method is a ready-made tool that belongs to a string. We use it by writing a dot after the string, then the method name and brackets, like `txt.upper()`.

Very important: strings cannot be changed once made. So a method never changes the original string, it gives us a **new** string. If we want to keep the new one, we must save it in a variable, like `txt = txt.upper()`.

````python
# upper() makes every letter CAPITAL and gives a new string
# (the last line of a cell is shown automatically, so no print() needed here)
txt.upper()
````

````python
# lower() makes every letter small
# we can use a method directly on a string written in quotes also, no variable needed
"MY NAME IS JOHN".lower()
````

````python
# title() makes the first letter of EVERY word capital, and the remaining letters small
txt.title()
````

````python
# capitalize() makes only the very first letter of the whole string capital, rest all small
# difference from title(): title() works on every word, capitalize() works only on the start of the sentence
txt.capitalize()
````

````python
# count() tells how many times a small piece of text (substring) is present in the string
# in "hello world" the letter o comes two times, so the answer is 2
txt.count("o")
````

````python
# replace(old, new) swaps the old text with the new text, everywhere it is found
# here every "o" becomes "x"
txt.replace("o" , "x")
````

````python
# count=1 means "replace only the first one"
# so only the first "o" (in hello) changes, and the "o" in world stays as it is
txt.replace("o", "x" , count=1)
````
