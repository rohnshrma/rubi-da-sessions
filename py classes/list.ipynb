# Collections
used to store multiple values of similar or different data types

- list : ordered, indexed, mutable (changable), allows duplicates
- tuple : ordered, indexed, immutable (unchangable), allows duplicates
- set : unordered, unindexed, immutable (unchangable), no allows duplicates 
- dictionary : ordered, unindexed, mutable (changable), no allows duplicates

# list 

`[ ]` are used to create a list , where items are separated using `,`


---
`ordered, indexed, mutable (changable), allows duplicates`




```python
marks = [23,56,67,34,67,89]
```

```python
print(type(marks) , len(marks))
```

```python
print(marks) # ordered and allows duplicates
```

```python
print(marks[0])
print(marks[2]) # indexed
```

```python
marks[3] = 64 # changable
```

```python
print(marks)
```

```python
# List methods
```

```python
# add a new item at the end of the list
marks.append(100) 
print(marks)
```

```python
# add a new item at the index specified
marks.insert(3 , 99) # insert index, item 
print(marks)
```

```python
# add multiple items at the end of the list

marks.extend( [56,56,56] ) # another list of items
print(marks)
```

```python
# remove the last item (by default) or the item on the index specified
marks.pop() #  last item
marks.pop(0) # item on the given index
print(marks)
```

```python
# remove first occurance of the item specified, raises an error if not found
marks.remove(56) 
print(marks)
```

```python
marks.remove(1000)
print(marks)
```

```python
# return number of occurance of the item specified
print(marks.count(56))
```

```python
# return index of the first occurance of the item specified
print(marks.index(56))
```

```python
# reverse order of the items
marks.reverse()
print(marks)
```

```python
# sort items asc or desc
marks.sort() # by default ascending
print(marks)
marks.sort(reverse=True) # descending
print(marks)
```

```python
x= marks.copy()
print(x)
```

```python
# empties a list
x.clear()
```
