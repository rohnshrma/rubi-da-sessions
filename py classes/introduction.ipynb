## Output
**print()** function is used to output data on screen

- `print()` is a *built-in function*: Python already knows it, you just call it using brackets `()`.
- Whatever you put **inside the brackets** (called the *argument*) is displayed.
- Text must be wrapped in quotes (`"..."` or `'...'`) , numbers and maths expressions are written without quotes.

**Example 1 : printing a calculation**

Python can work as a calculator. Here `2+2` is a maths expression: Python first **evaluates** it (gets `4`) and then `print()` shows the result. Notice there are no quotes, so Python treats it as maths and not as text.

Common operators: `+` add, `-` subtract, `*` multiply, `/` divide.

```python
print(2+2)
```

**Example 2 : printing text**

Text in Python is called a **string**. Because `"hello john"` is inside quotes, Python does *not* try to calculate it, it just displays it exactly as written.

Compare: `print(2+2)` shows `4` but `print("2+2")` would show `2+2`.

```python
print("hello john")
```

## Input
**input()** function is used to collect input from user

- When `input()` runs, the program **pauses** and waits for the user to type something and press `Enter`.
- The text inside the brackets is the **prompt**, a message telling the user what to type.
- `input()` **always gives back text (a string)**, even if the user types a number. Remember this, it matters later!

**Example : asking for a name**

The prompt `"Enter your name : "` is shown, then the user typed `JOHN`. The value typed is returned and displayed below the cell as `'JOHN'` (the quotes show it is a string).

Right now the answer is only displayed, **not saved anywhere**. To keep it, we need a variable (next section).

```python
input("Enter your name : ")
```

## Variable
empty containers used to store data, having a label on them known as variable name or identifier

Think of a variable like a **labelled box**: the label is the variable name and what is inside the box is the value. You can open the box any time (use the variable) or swap what is inside (assign a new value).

---

### How to create a variable

the moment you assign a value to a word it becomes a variable

- `=` is the **assignment operator**. It does *not* mean "equal to" as in maths. It means "store the value on the right into the name on the left".
- Python figures out the type of data automatically, you don't need to declare it.

**Example 1 : storing a number**

`a = 34` creates a variable called `a` and stores the number `34` in it. `print(a)` then shows the value stored, not the letter "a".

```python
a = 34
print(a)
```

**Example 2 : `None`**

`None` is a special value meaning **"no value yet" / "empty"**. It is useful when you want to create a variable now and fill it in later. It is not the same as `0` or an empty string.

This cell only creates the variable, so it shows no output.

```python
x = None
```

Printing `x` proves that it really holds `None`.

```python
print(x)
```

**Example 3 : changing a value**

Assigning again to the same name **replaces** the old value (`None` becomes `34`). A variable can be updated as many times as you like, and it always remembers the latest value.

```python
x = 34
```

Checking the new value of `x` after the reassignment.

```python
print(x)
```

**Example 4 : combining `input()` and a variable**

Here the user's answer is **saved** in the variable `age`, so we can use it later. Nothing is printed by the code itself, the only thing visible is the typed answer.

```python
age = input("Enter your age : ")
```

Printing `age` shows what the user typed.

> **Careful:** the value is `"34"` (text), not the number `34`, because `input()` always returns a string. So `age + 1` would give an error. Convert it first with `int(age)` (you will learn type conversion soon).

```python
print(age)
```

# variable naming conventions

Python only accepts certain names. Follow these rules or you get a `SyntaxError`:

<ul>
    <li>No Space &nbsp;(<code>my name</code> is wrong, <code>my_name</code> is right)</li>
    <li>A variable name cannot start with a number but can include or end with a number &nbsp;(<code>1st</code> is wrong, <code>name1</code> is right)</li>
    <li>A variable name cannot start , include or end with a special symbol except "_" &nbsp;(<code>my-name</code>, <code>name@</code> are wrong)</li>
    <li>Do Not Use Reserved Keywords &nbsp;(<code>if = 5</code> is wrong)</li>
    <li>Names are <b>case sensitive</b>: <code>age</code>, <code>Age</code> and <code>AGE</code> are three different variables</li>
</ul>

### Suggestions
- start variable name with lowercase letters only
- use meaningful names: `age` is better than `a`, `total_price` is better than `tp`
- in case of multiword variable name, e.g., `helloworldmynameis` is hard to read, so use one of these styles:
    - helloWorldMyNameIs : `camelCasing` (first word small, every next word starts with a capital)
    - hello_world_my_name_is : `snake_casing` (words separated by underscore, the Python standard)

**Example : a multiword variable name (camelCase)**

`yourName` follows camelCase. The line is split into three parts, explained in the comment cell at the bottom of this notebook: the name, the `=` operator and the value (a string, so it is in quotes).

```python
yourName = "john doe" 
```

**Example : reassigning and printing**

1. `a` starts as `None` (empty) and `print(a)` shows `None`.
2. `a` is reassigned to `100` and `print(a)` shows `100`.

This proves a variable holds only its **latest** value.

```python
a = None
print(a)
a = 100
print(a)
```

**Example : using variables in maths**

Variables can be used in calculations just like numbers. `a + a` becomes `100 + 100 = 200`. In a notebook, the **last line of a cell** is displayed automatically, so no `print()` was needed here.

```python
a + a
```
